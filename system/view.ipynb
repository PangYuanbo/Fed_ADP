```python
import re
import os
import numpy as np
import matplotlib.pyplot as plt

def read_accuracy(filename):
    rounds = []
    accs = []
    with open(filename, 'r') as f:
        for line in f:
            m = re.search(r"Round\s+(\d+):\s+ACC\s*=\s*([\d\.]+)", line)
            if m:
                rounds.append(int(m.group(1)))
                accs.append(float(m.group(2)))
    return rounds[1:], accs[1:]

def read_data(filename):
    grad_head, grad_feat = [], []
    with open(filename, 'r') as f:
        for line in f:
            mh = re.search(r"'grad_norm_head':\s*([\d\.]+)", line)
            mf = re.search(r"'grad_norm_feat':\s*([\d\.]+)", line)
            if mh and mf:
                grad_head.append(float(mh.group(1)))
                grad_feat.append(float(mf.group(1)))
    return grad_head, grad_feat

def view_data_avg(name, client_ids, num, ifdp=False):
    suffixes = ["train_after", "train_before", "test", "test_before"]
    # 存放所有客户端的数据
    grads_all = []    # list of dicts: { 'after_head': np.array, ... }
    accs_all = []     # list of np.array

    for cid in client_ids:
        # 构造日志路径前缀
        subdir = f"{name}_gradient_log{'_dp' if ifdp else ''}"
        base = os.path.join("logs", subdir,
                            f"gradient_log_client{cid}_{name}_{num}_0.0050.txt")

        # 读梯度
        d = {}
        for suf in suffixes:
            h, f = read_data(base + suf)
            d[f"{suf}_head"] = np.array(h)
            d[f"{suf}_feat"] = np.array(f)
        grads_all.append(d)

        # 读准确率（如果存在）
        acc_file = f"results/results_{name}_{cid}.txt"
        if os.path.exists(acc_file):
            _, acc = read_accuracy(acc_file)
            accs_all.append(np.array(acc))

    # 对齐并取平均——梯度
    # 先找出每条曲线在所有客户端里共同的最小长度
    min_len = {
        key: min(d[key].shape[0] for d in grads_all)
        for key in grads_all[0].keys()
    }
    avg_grads = {
        key: np.mean([d[key][:min_len[key]] for d in grads_all], axis=0)
        for key in grads_all[0].keys()
    }

    # 对齐并取平均——准确率
    if accs_all:
        min_acc = min(a.shape[0] for a in accs_all)
        avg_acc = np.mean([a[:min_acc] for a in accs_all], axis=0)
    else:
        avg_acc = None

        # —— 画平均梯度范数（拆分成 head 和 feat 两图） —— #

    # 提取所有 head 和 feat 的 key
    head_keys = [k for k in avg_grads if "_head" in k]
    feat_keys = [k for k in avg_grads if "_feat" in k]

    # ---- 画 head 梯度范数 ---- #
    plt.figure(figsize=(10, 5))
    for key in head_keys:
        series = avg_grads[key]
        plt.plot(range(1, len(series)+1), series,
                 linestyle="-", linewidth=1,
                 label=key.replace("_", " "))
    plt.xlabel("Iteration")
    plt.ylabel("Average Head Gradient Norm")
    plt.title("Mean Gradient Norms (Head) Across Clients")
    plt.legend(ncol=2, loc="upper right")
    plt.grid(True)
    plt.tight_layout()

    # ---- 画 feat 梯度范数 ---- #
    plt.figure(figsize=(10, 5))
    for key in feat_keys:
        series = avg_grads[key]
        plt.plot(range(1, len(series)+1), series,
                 linestyle="--", linewidth=1,
                 label=key.replace("_", " "))
    plt.xlabel("Iteration")
    plt.ylabel("Average Feature Gradient Norm")
    plt.title("Mean Gradient Norms (Feature Extractor) Across Clients")
    plt.legend(ncol=2, loc="upper right")
    plt.grid(True)
    plt.tight_layout()


    # —— 画平均准确率 —— #
    # —— 画平均准确率 —— #
    if avg_acc is not None:
        plt.figure(figsize=(8, 4))
        plt.plot(range(1, len(avg_acc)+1), avg_acc, linewidth=2, label='Current Avg Accuracy')

        # 加载 baseline（cifar-10-normal）的平均准确率
        base_client_ids = list(range(10))
        base_accs_all = []
        for cid in base_client_ids:
            acc_file = f"results/results_cifar-10-normal_{cid}.txt"
            if os.path.exists(acc_file):
                _, acc = read_accuracy(acc_file)
                base_accs_all.append(np.array(acc))

        # 画 baseline（如果读取成功）
        if base_accs_all:
            min_base_acc = min(a.shape[0] for a in base_accs_all)
            avg_base_acc = np.mean([a[:min_base_acc] for a in base_accs_all], axis=0)
            min_plot_len = min(len(avg_acc), len(avg_base_acc))
            plt.plot(range(1, min_plot_len+1), avg_base_acc[:min_plot_len],
                     linestyle='--', linewidth=2, label='Baseline: cifar-10-normal')

        plt.xlabel("Round")
        plt.ylabel("Average Accuracy")
        plt.title("Mean Accuracy Across Clients")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()


    plt.show()


```

```python
name    = "cifar-10-dp"
client_ids = list(range(10))
num     = 1000
ifdp    = True

view_data_avg(name, client_ids, num, ifdp)
name    = "cifar-10-normal"
client_ids = list(range(10))
num     = 1000
ifdp    = False

view_data_avg(name, client_ids, num, ifdp)
```
